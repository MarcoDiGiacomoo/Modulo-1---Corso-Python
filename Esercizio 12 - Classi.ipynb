```python
#Crea una classe Persona con attributo “nomeˮ e metodo presentati().
#Poi crea una sottoclasse Studente che aggiunge lʼattributo “corsoˮ 
#e lo include nella presentazione. Infine, 
#rendi lʼattributo nome privato e permetti di leggerlo solo tramite un metodo dedicato

class persona:
    def __init__(self, nome):
        self.__nome = nome

    def get_nome(self):
        return self.__nome

    def presentati(self):
        print(f"Ciao, mi chiamo {self.__nome}")


class studente(persona):
    def __init__(self, nome, corso):
        super().__init__(nome)
        self.corso = corso

    def presentati(self):
        print(f"{self.get_nome()} corso {self.corso}")

giulia = studente("Giulia", "Informatica")
giulia.presentati()

```

```python
#Crea una classe Contatore.
#Attributo:
#valore = 0
#Metodo:
#incrementa()
#che aumenta il contatore di 1.

class contatore:
    def __init__(self, valore):
        self.valore = valore

    def incrementa(self):
        self.valore += 1
        print(self.valore)

c = contatore(5)
c.incrementa()
```

```python
#Classe Rettangolo.
#Attributi:
#base
#altezza
#Metodo:
#area()
#che restituisce:
#base * altezza

class rettangolo:
    def __init__(self, base, altezza):
        self.base = base
        self.altezza = altezza

    def calcolo_area(self):
        print(self.base * self.altezza)

rt = rettangolo(8, 4)
rt.calcolo_area()
```

```python
#Attributi:
#nome
#corso
#Implementa:
#__str__()
#In modo che:
#print(studente)
#stampi:
#Giulia - Informatica

class studente:
    def __init__(self, nome, corso):
        self.nome = nome
        self.corso = corso

    def __str__(self):
        return self.nome + " - " + self.corso

giulia = studente("Giulia", "Informatica")
print(giulia)
```

```python
#Classe Auto.
#Attributi:
#marca
#modello
#anno
#Con __str__() che restituisce:
#Fiat Panda (2020)

class auto:
    def __init__(self, marca, modello, anno):
        self.marca = marca
        self.modello = modello
        self.anno = anno

    def __str__(self):
        return (self.marca + " " + self.modello + " (" + self.anno + ")")

panda = auto("Fiat", "Panda", "2020")
punto = auto("Fiat", "Punto", "2005")
print(panda)
print(punto)
```

```python
#Classe Film.
#Implementa un __str__() ben formattato.
#Ad esempio:
#Interstellar - Christopher Nolan

class film:
    def __init__(self, titolo, regista):
        self.titolo = titolo
        self.regista = regista
    def __str__(self):
        return (self.titolo + " - " + self.regista)

flm = film("Interstellar", "Christopher Nolan")
print(flm)
```

```python
#Classe ContoBancario.
#Attributo privato:
#__saldo
#Metodo:
#get_saldo()
#per leggere il saldo.

class conto:
    def __init__(self, saldo):
        self.__saldo = saldo

    def get_saldo(self):
        return self.__saldo

    def deposita(self, dep):
        self.__saldo += dep
        return self.__saldo

    def preleva(self, prl):
        if prl <= self.__saldo:
            self.__saldo -= prl
            return self.__saldo
        else:
            print("Prelievo non permesso")

saldo = conto(1890)

saldo.deposita(500)
saldo.preleva(300)
saldo.get_saldo()
```
