```python
#Esercizio 4
#Data la lista:
#numeri = [4, 8, 2, 9, 1]
#Calcola:
#somma
#numero più grande
#numero più piccolo
#(Senza usare max() e min(), se vuoi una sfida.)

numeri = [4, 8, 2, 9, 1]
somma = 0
minimo = numeri[0]
massimo =  numeri[0]

for i in numeri:
    somma += i

for n in numeri:
    if minimo > n:
        minimo = n

for m in numeri:
    if massimo < m:
        massimo = m

print(f"Somma: {somma}")
print(f"Minimo: {minimo}")
print(f"Massimo: {massimo}")
```

```python
#Realizza una rubrica telefonica.

#Il programma deve:

#Creare un dizionario vuoto.
#Permettere all'utente di inserire:
#nome
#numero
#Salvare il tutto nel dizionario.
#Continuare finché l'utente scrive "si".
#Alla fine stampare tutta la rubrica.

#Esempio:

#Python
#{
#"Marco": "3331234567",
#"Luca": "3499876543",
#"Anna": "3201112233"
#}

rub = {}
richiesta = input("Vuoi inserire un nuovo contatto in rubrica? Si/No: ")
while richiesta.lower() == "si":
    nome = input("Inserisci Nome: ")
    num = input("Inserisci numero: ")
    rub[nome] = num
    richiesta = str(input("Vuoi inserire un nuovo contatto in rubrica? Si/No: "))
print(rub)
```
