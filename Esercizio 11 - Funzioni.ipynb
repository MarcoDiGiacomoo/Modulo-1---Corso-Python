```python
#Crea una funzione:
#somma(a, b)
#che restituisce la somma usando return.
#Poi salva il risultato in una variabile e stampalo.

def somma(a, b):
    c = a + b
    return c

m = float(input("Inserisci un numero:"))
n = float(input("Inserisci un altro numero:"))
s = somma(m, n)
print("La somma dei due numeri inseriti è:", s)
```

```python
#Crea una funzione:
#moltiplica(a, b)
#che restituisce il prodotto dei due numeri.

def moltiplica(a, b):
    pr = a * b
    return pr

m = float(input("Inserisci un numero:"))
n = float(input("Inserisci un altro numero:"))
print("Il prodotto dei due numeri inseriti è:", moltiplica(m, n))
```

```python
#Crea una funzione:
#maggiore(a, b)
#che restituisce il numero più grande fra i due.
#Senza usare max().
def maggiore (a, b):
    if a > b:
        return a
    elif a == b:
        print("I due numeri sono uguali")
    else:
        return b

m = float(input("Inserisci un numero:"))
n = float(input("Inserisci un altro numero:"))
print("Il maggiore tra i due numeri è:", maggiore(m, n))
```

```python
#Crea una funzione:
#pari(numero)
#che restituisce:
#True
#se il numero è pari, altrimenti:
#False

def pari(x):
    if x % 2 == 0:
        return True
    else:
        return False

n = int(input("Inserisci un numero"))
print(f"{n} è pari?", pari(n))
```

```python
#Crea una funzione:
#voto(v)
#che restituisce:
#"Insufficiente" se < 6
#"Sufficiente" se = 6
#"Buono" se tra 7 e 8
#"Ottimo" se ≥ 9

def voto(x):
    if x < 6:
        print("Insufficiente")
    elif x == 6:
        print("Sufficiente")
    elif x == 7 or x == 8:
        print("Buono")
    else:
        print("Ottimo")

v = int(input("Qual è il tuo voto?"))
voto(v)
```

```python
#Crea una funzione:
#somma_lista(lista)
#che riceve una lista di numeri e restituisce la somma totale.

def somma_lista(lista):
    somma = 0
    for i in lista:
        somma += i
    return somma

i = 0
lista = []

while i < 5:
    n = int(input("Inserisci un numero:"))
    lista.append(n)
    i += 1

print(lista)
print("La somma dei numeri inseriti è:", somma_lista(lista))
```

```python
#Crea una funzione:
#trova_minimo(lista)
#che restituisce il numero più piccolo della lista senza usare min().

def trova_minimo(lista):
    minimo = lista[0]
    for i in lista:
        if i < minimo:
            minimo = i
    return minimo

lista = []
i = int(input("quanti numeri vuoi inserire nella lista?"))
for m in range(0, i):
    n = int(input("Inserisci un un numero:"))
    lista.append(n)

print(lista)
print("Il più piccolo tra i numeri inseriti è:", trova_minimo(lista))
```
