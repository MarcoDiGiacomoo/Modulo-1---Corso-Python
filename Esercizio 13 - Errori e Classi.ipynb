```python
#Chiedi due numeri e stampa la divisione.
#Gestisci:
#lettere inserite al posto dei numeri
#divisione per zero

try:
    n1 = int(input("Inserisci un numero qualsiasi"))
    n2 = int(input("Inserisci un numero qualsiasi"))
    div = n1 / n2
    print(div)
except ValueError:
    print("I valori inseriti non sono due interi")
except ZeroDivisionError:
    print("La divisione per 0 crea problemi")
```

```python
#Scrivi un programma che:
#chiede due numeri
#divide il primo per il secondo

try:
    a = int(input("Inserisci un numero"))
    b = int(input("Inserisci un altro numero"))
    div = a / b
except ValueError:
    print("È stata inserita una stringa")
except ZeroDivisionError:
    print("Non puoi dividere per 0")
else:
    print("Il risultato della divisione è: ", div)
finally:
    print("Fine Programma!")
```

```python
#Crea una classe:
#Persona
#con attributi:
#nome
#età
#e il metodo:
#presentati()
#Output:
#Ciao, sono Marco e ho 20 anni.

class persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def presentati(self):
        print(f"Ciao, sono {self.nome} e ho {self.eta} anni")

marco = persona("Marco", 20)
marco.presentati()
```

```python
#Crea una classe:
#Studente
#che eredita da Persona.
#Aggiungi:
#corso
#Modifica presentati().
#Output:
#Ciao, sono Giulia e studio Informatica.

class persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

class studente(persona):
    def __init__(self, nome, eta, corso):
        super().__init__(nome, eta)
        self.corso = corso

    def presentati(self):
        print(f"Ciao sono {self.nome}, frequento il corso di {self.corso}")

giulia = studente("Giulia", 20, "Informatica")
giulia.presentati()
```

```python
#Classe:
#ContoBancario
#Attributo:
#saldo
#Metodo:
#preleva(importo)
#Se l'importo supera il saldo:
#raise ValueError("Saldo insufficiente")

class ContoBancario:
    def __init__(self, saldo):
        self.saldo = saldo

    def preleva(self):
        importo = int(input("Inserisci l'importo da prelevare:"))
        if importo > self.saldo:
            raise ValueError("Saldo Insufficiente")
        else:
            self.saldo -= importo
            print(f"Importo prelevato {importo}, saldo {self.saldo}")

conto = ContoBancario(2300)
conto.preleva()
```

```python
#Crea:
#class SaldoInsufficienteError(Exception):
#pass
#Usala nella classe ContoBancario invece di ValueError.

class SaldoInsufficienteError(Exception):
    pass

class ContoBancario:
    def __init__(self, saldo):
        self.saldo = saldo

    def preleva(self):
        importo = int(input("Inserisci l'importo da prelevare:"))
        if importo > self.saldo:
            raise SaldoInsufficienteError("L'importo da prelevare supera il saldo")
        else:
            self.saldo -= importo
            print(f"Prelevati: {importo}, Saldo {self.saldo}")

conto = ContoBancario(2500)
conto.preleva()
```

```python
#Esercizio - Biblioteca
#Classe:
#Libro
#Attributi:

#titolo
#autore
#copie

#Se le copie sono negative:
#raise ValueError

#Metodo:
#presta_libro()

#Se non ci sono copie disponibili:
#raise Exception("Libro non disponibile")

class libro:
    def __init__(self, titolo, autore, copie):
        self.titolo = titolo
        self.autore = autore
        self.copie = copie
        
        if self.copie < 0:
            raise ValueError("Errore: numero negativo di copie")

    def presta_libro(self):
        if self.copie < 1:
            raise Exception("Libro non disponibile")
        else:
            self.copie -= 1
            print("Copie rimaste: ", self.copie)

book = libro("1984", "Orwell", 4)
book.presta_libro()

        
```
